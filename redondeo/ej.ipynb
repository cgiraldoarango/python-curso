# Redondeo

Formato general
round(numero, ndigits)
* numero: el valor numérico que quieres redondear
* ndigits: la cantidad de decimales a redondear (opcional)

```python
numero = 23.5677
print(round(numero, 2))
```

```python
print(round(numero,3))
```

```python
print(97 / 7)
# redondeo
print(round(97 / 7))
print(round(97 / 7, 4))  # cuatro decimales
print(round(97 / 7, 3))  # tres decimales
```

```python
nombre = input("ingresa tu nombre")
ventas = float(input("ingresa el valor de las ventas"))
comision = ventas * 1.3/100
comision = round(comision,2)
print(f"{nombre}, tu comision es: ${comision}")
```
