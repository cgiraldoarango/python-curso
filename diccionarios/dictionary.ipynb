# Diccionarios en Python

Los diccionarios son una de las estructuras de datos más importantes y utilizadas de Python. Permiten almacenar información mediante pares clave: valor, facilitando búsquedas rápidas y organización de datos. Son una estructura mutable y forman parte de los tipos de datos integrados del lenguaje

### 1. ¿Qué es un diccionario?
Un diccionario es una colección de pares:
+ clave: valor

```python
persona = {
    "nombre": "Carlos",
    "edad": 40,
    "ciudad": "Buga"
}
```

### 2. Características de los diccionarios
Son mutables

#### Podemos:
+ Agregar elementos.
+ Modificar elementos.
+ Eliminar elementos.

Ejemplo:

```python
persona = {
    "nombre": "Carlos"
}

persona["edad"] = 40

print(persona)
```

No permiten claves duplicadas

```python
datos = {
    "nombre": "Carlos",
    "nombre": "Juan"
}
```

La última definición reemplaza las anteriores.

Las claves deben ser hashables\
✅ Permitidas:\
"texto"\
123\
3.14\
True\
(1, 2)

❌ No permitidas:\
[]\
{}\
set()

Ejemplo incorrecto:

```python
dic = {
    [1, 2]: "valor"
}
```

Las claves deben ser objetos inmutables y hashables.\
[!WARNING]
**Hashables:** En programación y específicamente en Python, un objeto es hashable (o hasheable) si tiene un valor de hash que nunca cambia durante su vida útil y puede ser comparado con otros objetos.

### 3. Cómo crear diccionarios
Forma 1: llaves

```python
persona = {}
```

Forma 2: constructor dict()

```python
persona = dict()
```

Ambas crean un diccionario vacío.

Forma 3: con datos iniciales

```python
persona = {
    "nombre": "Carlos",
    "edad": 40
}
```

Forma 4: dict con argumentos

```python
persona = dict(
    nombre="Carlos",
    edad=40
)

print(persona)
```

Forma 5: usando una lista de tuplas

```python
persona = dict([
    ("nombre", "Carlos"),
    ("edad", 40)
])
print(persona)
```

### 4. Acceso a elementos

```python
persona = {
    "nombre": "Carlos"
}

print(persona["nombre"])
```

Error por clave inexistente

```python
print(persona["apellido"])
```

### Método get()

Más seguro:

```python
print(persona.get("apellido"))
```

Con valor por defecto:

```python
print(persona.get("apellido", "No existe"))
```

### 5. Agregar elementos

```python
persona = {}

persona["nombre"] = "Carlos"
persona["edad"] = 40

print(persona)
```

### 6. Modificar elementos

```python
persona = {
    "edad": 40
}

persona["edad"] = 41
```

### 7. Eliminar elementos
#### del

```python
persona = {
    "nombre": "Carlos",
    "edad": 40
}

del persona["edad"]

print(persona)
```

#### pop()

Devuelve el valor eliminado:

```python
persona={'nombre':'pablo',
         'ciudad':'cali',
         'edad':23}
edad = persona.pop("edad")
print(edad)
print(persona)
```

### popitem()
Elimina el último elemento insertado.

```python
persona.popitem()
```

### clear()
Vacía el diccionario.

```python
persona.clear()
```

### 8. Método keys()
Devuelve las claves.

```python
persona = {
    "nombre": "Carlos",
    "edad": 40
}

print(persona.keys())
```

### 9. Método values()
Devuelve los valores.

```python
print(persona.values())
```

### 10. Método items()
Devuelve clave y valor.

```python
print(persona.items())
```

### 11. Recorrer diccionarios
Recorrer claves

```python
for clave in persona:
    print(clave)
```

Recorrer clave y valor

```python
for clave, valor in persona.items():
    print(clave, valor)
```

### 12. Operador in
Permite verificar si una clave existe.

```python
persona = {
    "nombre": "Carlos"
}

print("nombre" in persona)
print("edad" in persona)
```

### 13. Longitud de un diccionario

```python
persona = {
    "nombre": "Carlos",
    "edad": 40
}

print(len(persona))
```

### 14. Copiar diccionarios
Error común. No se copia.
Ambas variables apuntan al mismo objeto.

Copia correcta

```python
a ={'nombre':'pablo'}
b = a.copy()
```

### 15. Diccionarios anidados
Podemos almacenar otros diccionarios.

```python
persona = {
    "nombre": "Carlos",
    "direccion": {
        "ciudad": "Buga",
        "pais": "Colombia"
    }
}
```

Acceso

```python
print(persona["direccion"]["ciudad"])
```

### 16. Diccionarios con múltiples tipos
Un diccionario puede contener cualquier tipo de dato.

```python
datos = {
    "texto": "Hola",
    "entero": 10,
    "decimal": 3.14,
    "booleano": True,
    "lista": [1, 2, 3],
    "tupla": (10, 20),
    "diccionario": {
        "a": 1
    }
}
```

### 17. Método update()
Actualiza o agrega elementos.

```python
persona = {
    "nombre": "Carlos"
}

persona.update({
    "edad": 40,
    "ciudad": "Buga"
})

print(persona)
```

### 18. Método setdefault()
Si la clave no existe, la crea.
Python

```python
persona = {
    "nombre": "Carlos"
}

persona.setdefault("edad", 40)

print(persona)
```

### 19. Comprensión de diccionarios
Equivalente a las comprensiones de listas.

```python
cuadrados = {
    x: x**2
    for x in range(1, 6)
}

print(cuadrados)
```

### 20. Orden de los diccionarios
Desde Python 3.7 los diccionarios conservan el orden de inserción.\
Ejemplo:

```python
datos = {}

datos["a"] = 1
datos["b"] = 2
datos["c"] = 3

print(datos)
```

### 21. Operaciones comunes
Obtener todas las claves

```python
list(persona.keys())
```

Obtener todos los pares

```python
list(persona.items())
```

Convertir a lista

```python
list(persona)
```

### 22. Ejemplos prácticos
Contar palabras

```python
texto = "python python java python"

contador = {}

for palabra in texto.split():
    contador[palabra] = contador.get(palabra, 0) + 1

print(contador)
```

Agenda telefónica

```python
agenda = {
    "Carlos": "3001111111",
    "Ana": "3002222222"
}

nombre = "Carlos"

print(agenda[nombre])
```

Inventario

```python
inventario = {
    "teclados": 20,
    "ratones": 15,
    "monitores": 8
}

for producto, cantidad in inventario.items():
    print(f"{producto}: {cantidad}")
```

### Glosario
+ Clave (Key): identificador único del dato.
+ Valor (Value): información asociada a la  clave.
+ Hashable: objeto que puede utilizarse como clave.
+ Mutable: puede modificarse después de crearse.
+ Items: pares clave-valor.
+ Comprensión de diccionario: creación de diccionarios mediante expresiones compactas.
+ Diccionario anidado: diccionario que contiene otros diccionarios.
