# Conversiones entre tipos

Este tutorial está basado en los conceptos fundamentales de tipos de datos, conversiones y la función type() presentes en el material de referencia del cuaderno

### 1. ¿Qué es un tipo de dato?

En Python, todo objeto tiene un tipo. El tipo determina:
1. Qué valores puede almacenar.
2. Qué operaciones puede realizar.
3. Cómo se comporta en el programa.

Los tipos básicos más utilizados son:

|Tipo	| Descripción	|Ejemplo |
|-------|:---------------:|--------:|
|int	|Enteros	    |10, -5  |  
|float	|Decimales	    |3.14, 2.5|
str	|Texto	|"Hola"|
bool|	Booleanos|	True, False|
NoneType|	Valor nulo|	None|

### 2. Uso de la función type()

La función type() permite conocer el tipo de una variable.

```python
edad = 25
print(type(edad))
```

#### Más ejemplos

```python
numero = 5
decimal = 3.14
texto = "Python"
activo = True

print(type(numero))
print(type(decimal))
print(type(texto))
print(type(activo))
```

### 3. Conversiones entre tipos

Python permite convertir explícitamente un tipo a otro mediante funciones de conversión.

#### 3.1 Convertir a entero: int()

Convierte números o textos numéricos en enteros.

```python
numero = "100"

valor = int(numero)

print(valor)
print(type(valor))
```

Conversión de float a int

```python
precio = 19.99

entero = int(precio)

print(entero)
```

⚠️ int() elimina la parte decimal, no redondea.

#### 3.2 Convertir a decimal: float()

Convierte enteros o textos numéricos a números decimales.

Ejemplo

```python
numero = 20

decimal = float(numero)

print(decimal)
print(type(decimal))
```

Desde texto

```python
x = float("3.1416")

print(x)
```

#### 3.3 Convertir a texto: str()

Transforma cualquier valor a cadena de texto.

Ejemplo

```python
edad = 25

texto = str(edad)

print(texto)
print(type(texto))
```

Aplicación práctica

Incorrecto

```python
edad = 25

print("Mi edad es: " + edad)
```

Correcto

```python
edad = 25

print("Mi edad es: " + str(edad))
```

#### 3.4 Convertir a booleano: bool()

Convierte valores a True o False.

Ejemplos

```python
print(bool(1))
print(bool(0))
```

```python
print(bool("Hola"))
print(bool(""))
```

### 4. El caso especial de input()

Un error muy frecuente es olvidar que input() siempre devuelve un string.

Ejemplo incorrecto

```python
num1 = input("Número 1: ")
num2 = input("Número 2: ")

print(num1 + num2)
```

Solución

```python
num1 = int(input("Número 1: "))
num2 = int(input("Número 2: "))

print(num1 + num2)
```

### 5. Tabla resumen de conversiones

|Función	|Convierte a
|-----------|-----------|
|int()	|Entero
|float()	|Decimal
|str()	|Texto
|bool()	|Booleano
|type()	|Consulta tipo

### 6. Ejemplo completo

```python
valor = input("Ingrese un número: ")

print("Valor original:", valor)
print("Tipo original:", type(valor))

entero = int(valor)
decimal = float(valor)
texto = str(valor)
booleano = bool(valor)

print("\nConversiones:")
print(entero, type(entero))
print(decimal, type(decimal))
print(texto, type(texto))
print(booleano, type(booleano))
```

Errores comunes

1. Convertir texto no numérico a entero

```python
int("hola")
```

2. Mezclar texto con números

```python
"Edad: " + 25
```

3. Creer que input() devuelve números

```python
numero = input("Número: ")
```

Debes convertir:

```python
numero = int(input("Número: "))
```

### Glosario

1. Casting: Conversión explícita de un tipo a otro.
2. int: Número entero.
3. float: Número decimal.
4. str: Cadena de texto.
5. bool: Valor lógico (True o False).
6. type(): Función que retorna el tipo de un objeto.
7. input(): Función que captura datos desde teclado y devuelve un string.
