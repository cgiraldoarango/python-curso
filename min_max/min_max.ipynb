# Funciones min() y max()
Las funciones integradas min() y max() son herramientas fundamentales en Python que permiten encontrar, respectivamente, el valor mínimo y el valor máximo dentro de una colección de datos o entre varios valores. Forman parte de las funciones incorporadas del lenguaje.

### 1. Definición de min()

La función min() devuelve el elemento más pequeño de un conjunto de valores o de un iterable (lista, tupla, cadena, etc.).

Sintaxis\
min(iterable)

#### Ejemplos básicos
Con varios números

```python
resultado = min(10, 5, 20, 3)
print(resultado)
```

Con una lista

```python
numeros = [8, 4, 2, 9, 1]
print(min(numeros))
```

### 2. Definición de max()

La función max() devuelve el elemento más grande de una colección o conjunto de valores.

Sintaxis\
max(iterable)

max(valor1, valor2, valor3, ...)

#### Ejemplos básicos
Con varios números

```python
resultado = max(10, 5, 20, 3)
print(resultado)
```

Con listas

```python
numeros = [8, 4, 2, 9, 1]
print(max(numeros))
```

### 3. Uso con cadenas de texto

Python compara cadenas según el orden Unicode/ASCII.

```python
palabras = ["casa", "árbol", "perro", "gato"]

print(min(palabras))
print(max(palabras))
```

### 4. Uso con tuplas

```python
datos = (45, 12, 78, 3, 95)

print(min(datos))
print(max(datos))
```

### 5. Uso con diccionarios

Cuando se aplican directamente a un diccionario, trabajan sobre las claves.

```python
edades = {
    "Juan": 20,
    "Ana": 18,
    "Pedro": 25
}

print(min(edades))
print(max(edades))
```

### 6. Uso con cadenas (string)

```python
texto = "python"

print(min(texto))
print(max(texto))
```

Python evalúa cada carácter individualmente.

### 7. Uso práctico: notas de estudiantes

```python
notas = [4.5, 3.8, 5.0, 2.9, 4.7]

nota_minima = min(notas)
nota_maxima = max(notas)

print("Nota más baja:", nota_minima)
print("Nota más alta:", nota_maxima)
```

### 8. Uso con el parámetro key

Una de las características más potentes de min() y max() es el parámetro key, que permite indicar el criterio de comparación.

Encontrar la palabra más larga

```python
palabras = ["sol", "computadora", "luna", "python"]

print(max(palabras, key=len))
```

Encontrar la palabra más corta

```python
palabras = ["sol", "computadora", "luna", "python"]

print(min(palabras, key=len))
```

### 9. Uso con listas de diccionarios

Supongamos que tenemos estudiantes:

```python
estudiantes = [
    {"nombre": "Ana", "nota": 4.5},
    {"nombre": "Carlos", "nota": 5.0},
    {"nombre": "Juan", "nota": 3.8}
]
```

Estudiante con mayor nota

```python
mejor = max(estudiantes, key=lambda e: e["nota"])

print(mejor)
```

Estudiante con menor nota

```python
peor = min(estudiantes, key=lambda e: e["nota"])

print(peor)
```

### 10. Casos prácticos de uso

Temperaturas

```python
temperaturas = [28, 31, 25, 22, 35, 29]

print("Temperatura mínima:", min(temperaturas))
print("Temperatura máxima:", max(temperaturas))
```

Ventas

```python
ventas = [1500, 2200, 1800, 3500, 1200]

print("Menor venta:", min(ventas))
print("Mayor venta:", max(ventas))
```

Edades

```python
edades = [18, 24, 35, 12, 67]

print("Edad mínima:", min(edades))
print("Edad máxima:", max(edades))
```

### Errores comunes

1. Lista vacía

```python
numeros = []

print(min(numeros))
```

Solución

```python
if numeros:
    print(min(numeros))
```

2. Mezclar tipos incompatibles

```python
datos = [1, "hola", 3]

print(max(datos))
```

Python no puede comparar enteros con cadenas.

Comparación rápida

|Función|	Propósito|	Resultado|
|-|-|-|
|min()|	Obtener el menor valor|	Valor mínimo|
|max()	|Obtener el mayor valor	|Valor máximo|

### Resumen

- min() devuelve el elemento más pequeño.
- max() devuelve el elemento más grande.
- Funcionan con listas, tuplas, cadenas, diccionarios y cualquier iterable.
- Admiten múltiples argumentos.
- El parámetro key permite personalizar el criterio de comparación.
- Son muy utilizadas en análisis de datos, estadísticas, reportes y validaciones.

### Glosario

- Iterable: objeto que puede recorrerse elemento a elemento (listas, tuplas, cadenas, etc.).
- Key: función utilizada para definir el criterio de comparación.
- Lambda: función anónima de una línea utilizada frecuentemente junto con key.
- ValueError: excepción lanzada cuando min() o max() reciben una secuencia vacía.
