# Tuplas en Python
Las tuplas son una de las estructuras de datos fundamentales de Python. Se utilizan para almacenar colecciones de valores de forma ordenada y tienen una característica muy importante: son inmutables, es decir, una vez creadas no pueden ser modificadas.

### 1. ¿Qué es una tupla?
Una tupla es una colección ordenada de elementos que puede contener diferentes tipos de datos.

```python
persona = ("Carlos", 35, True)
```

Las tuplas permiten almacenar datos heterogéneos en una sola estructura.

### 2. Cómo crear tuplas
#### Tupla básica

```python
frutas = ("manzana", "pera", "uva")
```

#### Tupla numérica

```python
numeros = (10, 20, 30)
```

#### Tupla mixta

```python
datos = ("Carlos", 40, 1.75, True)
```

#### Tupla vacía

```python
vacia = ()
```

#### Tupla de un solo elemento
⚠ Error común:

```python
tupla = (10)
print(type(tupla))
```

Forma correcta:

```python
tupla = (10,)
print(type(tupla))
```

### 3. Constructor tuple()

Python incorpora el constructor tuple() para crear tuplas a partir de otros iterables.
#### Desde una lista

```python
lista = [1, 2, 3, 4]

tupla = tuple(lista)

print(tupla)
```

#### Desde una cadena

```python
texto = "Python"

tupla = tuple(texto)

print(tupla)
```

### 4. Tipos de datos que puede contener una tupla
Una tupla puede contener prácticamente cualquier objeto Python.

```python
ejemplo = (
    100,            # int
    3.14,           # float
    "Python",       # str
    True,           # bool
    [1, 2, 3],      # list
    {"a": 1},       # dict
    (4, 5)          # tuple
)
```

Incluso puede contener otras estructuras complejas.

### 5. Indexado

Las posiciones comienzan en 0.

```python
colores = ("rojo", "verde", "azul")

print(colores[0])
```

#### Índices negativos

Permiten acceder desde el final.

```python
print(colores[-1])
```

```python
print(colores[-2])
```

### 6. Slicing (rebanado)
Permite obtener partes de una tupla.

```python
numeros = (10, 20, 30, 40, 50)

print(numeros[1:4])
```

Desde el inicio

```python
print(numeros[:3])
```

Hasta el final

```python
print(numeros[2:])
```

### 7. Propiedades de las tuplas

Las tuplas son:

✅ Ordenadas\
✅ Permiten duplicados\
✅ Indexables\
✅ Iterables\
✅ Más ligeras en memoria que las listas\
✅ Inmutables

### 8. Mutabilidad e inmutabilidad
Tuplas: inmutables

```python
tupla = (1, 2, 3)

tupla[0] = 100
```

Porque una tupla no puede modificarse después de creada

Caso especial\
La tupla es inmutable, pero si contiene objetos mutables estos sí pueden cambiar.

```python
tupla = ([1, 2, 3], "Python")

tupla[0].append(4)

print(tupla)
```

### 9. Tuplas anidadas
Una tupla puede contener otras tuplas.

```python
matriz = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)
```

Acceso

```python
print(matriz[1])
```

```python
print(matriz[1][2])
```

### 10. Operaciones con tuplas
Concatenación

```python
t1 = (1, 2, 3)
t2 = (4, 5)

resultado = t1 + t2

print(resultado)
```

Repetición

```python
t = ("Python",)

print(t * 3)
```

### 11. Recorrer tuplas
Con for

```python
lenguajes = ("Python", "Java", "Go")

for lenguaje in lenguajes:
    print(lenguaje)
```

Con índices

```python
for i in range(len(lenguajes)):
    print(i, lenguajes[i])
```

### 12. Desempaquetado de tuplas
Permite asignar elementos a variables.

```python
persona = ("Carlos", 35, "Colombia")

nombre, edad, pais = persona

print(nombre)
print(edad)
print(pais)
```

Intercambio de variables

```python
a = 10
b = 20

a, b = b, a

print(a, b)
```

### 13. Métodos de las tuplas
Las tuplas poseen solamente dos métodos propios.

#### count()
Cuenta cuántas veces aparece un valor.

```python
datos = (1, 1, 2, 3, 1)

print(datos.count(1))
```

####  index()
Devuelve la posición de la primera aparición.

```python
datos = (10, 20, 30, 40)

print(datos.index(30))
```

### 14. Funciones útiles para trabajar con tuplas
#### len()

```python
numeros = (1, 2, 3, 4)

print(len(numeros))
```

#### max()

```python
print(max((4, 10, 2)))
```

#### min()

```python
print(min((4, 10, 2)))
```

#### sum()

```python
print(sum((1, 2, 3, 4)))
```

### 15. Conversión entre listas y tuplas
Lista → Tupla

```python
lista = [1, 2, 3]

tupla = tuple(lista)
```

Tupla → Lista

```python
tupla = (1, 2, 3)

lista = list(tupla)
```

### 16. Usos más importantes de las tuplas
##### 1. Devolver múltiples valores desde funciones

```python
def obtener_persona():
    return ("Carlos", 35)

nombre, edad = obtener_persona()
```

##### 2. Coordenadas

```python
punto = (10, 20)
```

##### 3. Registros que no deben modificarse

```python
configuracion = (
    "localhost",
    5432,
    "postgres"
)
```

##### 4. Claves de diccionarios

Las listas no pueden ser claves porque son mutables.

```python
ubicaciones = {
    (10, 20): "Casa",
    (30, 40): "Oficina"
}
```

### Errores comunes
Olvidar la coma en una tupla de un elemento

```python
x = (10)
print(type(x))
```

No es una tupla.

Correcto

```python
x = (10,)
print(type(x))
```

Intentar modificar una tupla

```python
t = (1, 2, 3)

t[0] = 10
```

Confundir listas con tuplas
Python

```python
lista = [1, 2, 3]
tupla = (1, 2, 3)
```

* Lista → mutable
* Tupla → inmutable
