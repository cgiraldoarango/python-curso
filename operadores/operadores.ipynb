Operadores de Python

Los operadores son símbolos o palabras reservadas que permiten realizar operaciones sobre variables y valores

1. Operadores aritméticos

Operador	Descripción	Ejemplo
1. (+)	Suma	5 + 3 → 8
2. (-)	Resta	5 - 3 → 2
3. (*)	Multiplicación	5 * 3 → 15
4. (/)	División	5 / 2 → 2.5
5. (//)	División entera	5 // 2 → 2
6. (%)	Módulo (resto)	5 % 2 → 1
7. (**)	Potencia	2 ** 4 → 16

Ejemplo

```python
a = 10
b = 3

print(a + b)   # 13
print(a - b)   # 7
print(a * b)   # 30
print(a / b)   # 3.333...
print(a // b)  # 3
print(a % b)   # 1
print(a ** b)  # 1000
```

2. Operadores de comparación

Devuelven True o False.

Operador	Descripción
1. (==)	Igual
2. (!=)	Distinto
3. (>)	Mayor que
4. (<)	Menor que
5. (>=)	Mayor o igual que
6. (<=)	Menor o igual que

Ejemplo

```python
a = 5
b = 3

print(a == b)  # False
print(a != b)  # True
print(a > b)   # True
print(a < b)   # False
print(a >= b)  # True
print(a <= b)  # False
```

3. Operadores lógicos

Trabajan con valores booleanos.

Operador  Descripción
1. and	Y lógico
2. or	O lógico
3. not	Negación

Ejemplo

```python
x = True
y = False

print(x and y)  # False
print(x or y)   # True
print(not x)    # False
```

4. Operadores de asignación

Operador	Equivale a
1. (=)	Asignación
2. (+=)	x = x + valor
3. (-=)	x = x - valor
4. (*=)	x = x * valor
5. (/=)	x = x / valor
6. (//=)	x = x // valor
7. (%=)	x = x % valor
8. (**=)	x = x ** valor

Ejemplo

```python
x = 10

x += 5
print(x)  # 15

x *= 2
print(x)  # 30
```

5. Operadores de pertenencia

Permiten verificar si un elemento está dentro de una secuencia.

Operador	Descripción
1. (in)	Está contenido
2. (not in)	No está contenido

Ejemplo

```python
frutas = ["manzana", "pera", "uva"]

print("pera" in frutas)      # True
print("kiwi" not in frutas)  # True
```

6. Operadores de identidad

Comparan si dos variables apuntan al mismo objeto en memoria.

Operador	Descripción
1. (is)	Es el mismo objeto
2. (is not)	No es el mismo objeto

Ejemplo

```python
a = [1, 2]
b = a
c = [1, 2]

print(a is b)      # True
print(a is c)      # False
print(a == c)      # True
```

Glosario

* Operador: símbolo que realiza una operación.
* Operando: valor sobre el que actúa un operador.
* Booleano (bool): tipo de dato que puede ser True o False.
* Módulo (%): devuelve el resto de una división.
* Potencia (**): eleva un número a otro.
