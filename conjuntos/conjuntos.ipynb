# Conjuntos (set) en Python

Los conjuntos (sets) son estructuras de datos que permiten almacenar elementos únicos, sin duplicados y sin orden definido. Son muy útiles para eliminar repetidos, realizar operaciones matemáticas de teoría de conjuntos y hacer búsquedas rápidas.

### 1. ¿Qué es un conjunto?
Un conjunto es una colección de elementos:
- No ordenada.
- Mutable (se puede modificar).
- Sin elementos duplicados.
- Sus elementos deben ser hashables (inmutables).

```python
numeros = {1, 2, 3, 4}

print(numeros)
```

Si intentas agregar elementos repetidos:

```python
numeros = {1, 2, 3, 3, 3, 4}

print(numeros)
```

Los duplicados son eliminados automáticamente.

### 2. Constructor set()
La función set() permite crear conjuntos a partir de iterables.
- Desde una lista

```python
datos = [1, 2, 2, 3, 3, 4]

conjunto = set(datos)

print(conjunto)
```

- Desde una cadena

```python
texto = "python"

letras = set(texto)

print(letras)
```

Crear conjunto vacío

```python
conjunto = set()
```

⚠️ No debes usar:

```python
conjunto = {}
print(type(conjunto))
```

porque crea un diccionario vacío, no un conjunto.

### 3. Acceso a los elementos
Los conjuntos no tienen índices.
- Esto es incorrecto:

```python
numeros = {10, 20, 30}

print(numeros[0])
```

No se puede acceder por posición porque los conjuntos no están ordenados.

### 4. Cómo recorrer un conjunto
Se utilizan ciclos for.

```python
numeros = {10, 20, 30}

for numero in numeros:
    print(numero)
```

El orden puede variar.

### 5. Mutabilidad
- Los conjuntos son mutables.
#### Podemos agregar y eliminar elementos.
- Agregar

```python
numeros = {1, 2, 3}

numeros.add(4)

print(numeros)
```

Eliminar con remove

```python
numeros.remove(3)
```

Eliminar con discard

```python
numeros.discard(3)
```

#####  Diferencia:
- remove() genera error si no existe el elemento.
- discard() no genera error.
##### Vaciar conjunto

```python
numeros.clear()
```

### 6. Tipos de datos permitidos
Los elementos deben ser inmutables.
##### Permitidos

```python
{1, 2, 3}
```

```python
{"Juan", "Pedro"}
```

```python
{True, False}
```

```python
{1.5, 2.8}
```

```python
{(1, 2), (3, 4)}
```

No permitidos

```python
{[1, 2]}
```

Tampoco:

```python
{{1,2}}
```

porque un conjunto no puede contener otro conjunto mutable.

7. Conversión a otros tipos

Set → Tupla

```python
tupla = tuple(conjunto)
```

8. Verificar si un elemento está en un conjunto
##### Se usa el operador in.

```python
numeros = {1, 2, 3, 4}

print(3 in numeros)
```

```python
print(8 in numeros)
```

La búsqueda en conjuntos es muy eficiente.

9. Unir dos conjuntos
- Operador |

```python
a = {1, 2, 3}
b = {3, 4, 5}

union = a | b

print(union)
```

Método union()

```python
union = a.union(b)
```

### 10. Operaciones con conjuntos
Unión
##### Elementos de ambos conjuntos.

```python
A = {1, 2, 3}
B = {3, 4, 5}

print(A | B)
```

```python
print(A.union(B))
```

##### Intersección
Elementos comunes.

```python
print(A & B)
```

También

```python
A.intersection(B)
```

Diferencia
###### Elementos que están en A pero no en B.

```python
print(A - B)
```

Diferencia simétrica
- Elementos que no son comunes.

```python
print(A ^ B)
```

La documentación oficial de Python presenta estas operaciones como las operaciones fundamentales de conjuntos.

### 11. Relaciones entre conjuntos
Subconjunto

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))
```

Superconjunto

```python
print(B.issuperset(A))
```

#### Disjuntos
Sin elementos comunes.

```python
A = {1, 2}
B = {5, 6}

print(A.isdisjoint(B))
```

### 12. Propiedades de los conjuntos
Los conjuntos cumplen:
#### Unicidad

```python
{1, 1, 1, 2}
```

No tienen orden

```python
{1,2,3}
```

no garantiza ningún orden.

##### Búsqueda rápida

### 13. Restricciones
No tienen índices

```python
conjunto[0]
```

No permiten elementos mutables

```python
{[1,2]}
```

#### No mantienen orden de inserción
No debes asumir:

```python
{1, 2, 3}
```

se recorrerá siempre como:

No permiten duplicados

```python
{1, 1, 1, 2}
```

### Ejemplo completo

```python
estudiantes_python = {"Ana", "Juan", "Pedro"}
estudiantes_java = {"Juan", "Carlos", "Pedro"}

# Unión
print(estudiantes_python | estudiantes_java)

# Intersección
print(estudiantes_python & estudiantes_java)

# Diferencia
print(estudiantes_python - estudiantes_java)

# Verificar existencia
print("Ana" in estudiantes_python)

# Agregar
estudiantes_python.add("Luisa")

# Recorrer
for estudiante in estudiantes_python:
    print(estudiante)
```

### Glosario
- Set: Conjunto de elementos únicos.
- Hashable: Objeto que puede usarse dentro de un conjunto.
- Unión: Elementos de ambos conjuntos.
- Intersección: Elementos comunes.
- Subconjunto: Conjunto completamente contenido en otro.
- Diferencia simétrica: Elementos exclusivos de cada conjunto.
