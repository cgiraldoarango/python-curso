# Método index de los strings

### Formato general
cadena_o_lista.index(valor, inicio, fin)
* valor: el elemento o subcadena que buscas
* inicio: posición inicial desde donde buscar (opcional)
* fin: posición final hasta donde buscar (opcional)


```python
cadena = "Este es un mensaje de prueba"
print(cadena.index("e"))
print(cadena.index("e", 4))  # encuentra la segunda letra "e"
print(cadena.index("e", 6, 17))
```

## Slicing

```python
cadena = "Esta es una cadena de prueba"
dato = cadena[0]  # Índice cero corresponde a la primera posición
print(dato)
dato = cadena[3]
print(dato)
dato = cadena[2:8]  # no se incluye la posición 8
print(dato)
dato = cadena[2:17:2]  # step 2
print(dato)
dato = cadena[-1]  # ultimo elemento de la cadena
print(dato)
dato = cadena[-3]  # tercer elemento de atrás hacia adelante
print(dato)
dato = cadena[:]  # toma toda la cadena
print(dato)
dato = cadena[:11]  # desde el principio hasta la indice 10
print(dato)
dato = cadena[::-1]  # toda la cadena pero al revés
print(dato)
dato = cadena[-10:-2]  # cuenta desde atrás
print(dato)
dato = cadena[-10:-2:2]  # step 2
print(dato)
```
