# Método index de los strings

### Formato general
cadena_o_lista.index(valor, inicio, fin)
* valor: el elemento o subcadena que buscas
* inicio: posición inicial desde donde buscar (opcional)
* fin: posición final hasta donde buscar (opcional)


```python
cadena = "Este es un mensaje de prueba"
print(cadena.index("e"))  # cero corresponde a la posición del primer elemento
print(cadena.index("e", 4))  # encuentra la segunda letra "e"
```
