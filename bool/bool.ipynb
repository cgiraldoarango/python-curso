# Tipo de dato bool

El tipo de dato bool es uno de los más importantes en Python porque permite representar condiciones lógicas y controlar el flujo de ejecución de los programas mediante decisiones, validaciones y ciclos. Un valor booleano solo puede tener dos estados: True o False.

### 1. ¿Qué es un booleano (bool)?

Un booleano es un tipo de dato que representa una respuesta lógica:

```python
True
False
```

Podemos comprobar su tipo:

```python
print(type(True))
print(type(False))
```

#### Los valores booleanos son fundamentales para:
- Validaciones
- Comparaciones
- Condicionales (if)
- Ciclos (while)
- Filtrado de datos
- Control de acceso
- Verificación de estados

### 2. Crear valores booleanos directamente

```python
activo = True
bloqueado = False

print(activo)
print(bloqueado)
```

### 3. Crear booleanos mediante expresiones

La forma más común es mediante comparaciones.

Igual

```python
resultado = 5 == 5
print(resultado)
```

Distinto

```python
resultado = 5 != 3
print(resultado)
```

Mayor que

```python
resultado = 10 > 2
print(resultado)
```

Menor que

```python
resultado = 10 < 2
print(resultado)
```

Mayor o igual

```python
resultado = 10 >= 10
print(resultado)
```

Menor o igual

```python
resultado = 5 <= 2
print(resultado)
```

Las comparaciones producen automáticamente valores booleanos.

### 4. Operadores booleanos

Python incluye tres operadores lógicos principales.

AND

Devuelve True únicamente si ambas expresiones son verdaderas.

```python
edad = 25
tiene_documento = True

puede_entrar = edad >= 18 and tiene_documento

print(puede_entrar)
```

OR

Devuelve True si al menos una condición es verdadera.

```python
es_admin = False
es_supervisor = True

acceso = es_admin or es_supervisor

print(acceso)
```

NOT

Invierte el valor lógico.

```python
activo = True

print(not activo)
```

### 5. Validar si un dato existe usando in

El operador in devuelve un booleano.

En listas

```python
numeros = [10, 20, 30]

print(20 in numeros)
```

```python
print(50 in numeros)
```

En cadenas

```python
texto = "Python"

print("Py" in texto)
```

En tuplas

```python
colores = ("rojo", "verde", "azul")

print("verde" in colores)
```

En diccionarios

Busca en las claves:

```python
usuario = {
    "nombre": "Carlos",
    "edad": 30
}

print("nombre" in usuario)
```

### 6. Validar que un dato NO existe usando not in

```python
numeros = [1, 2, 3]

print(4 not in numeros)
```

Otro ejemplo:

```python
texto = "Python"

print("Java" not in texto)
```

### 9. Bool en estructuras condicionales
Python evalúa automáticamente expresiones booleanas.

```python
edad = 20

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
```

### 10. Usar bool en ciclos while
El tipo bool es muy común para controlar la ejecución de ciclos.
Python

```python
continuar = True

while continuar:
    print("Ejecutando")
    continuar = False
```

### 11. Evitar ciclos infinitos
Error común

```python
while True:
    print("Hola")
```

Este ciclo nunca finaliza.

Solución 1: Modificar la condición

```python
contador = 1

while contador <= 5:
    print(contador)
    contador += 1
```

Solución 2: Usar una bandera booleana

```python
ejecutando = True

while ejecutando:
    opcion = input("Salir (s/n): ")

    if opcion == "s":
        ejecutando = False
```

Solución 3: Usar break

```python
while True:

    opcion = input("Salir (s/n): ")

    if opcion == "s":
        break
```

### 13. Casos de uso reales
Validar usuario

```python
usuario_activo = True

if usuario_activo:
    print("Bienvenido")
```

Comprobar permisos

```python
es_admin = False

if not es_admin:
    print("Acceso denegado")
```

Validar datos obligatorios

```python
nombre = "Carlos"

if nombre:
    print("Nombre válido")
```

Validar existencia en una lista

```python
roles = ["admin", "editor"]

if "admin" in roles:
    print("Tiene privilegios")
```

Buscar elementos

```python
archivo = "datos.csv"

if archivo.endswith(".csv"):
    print("Archivo válido")
```

### Glosario
- Bool: Tipo lógico que almacena True o False.
- Expresión booleana: Expresión que devuelve un booleano.
- Operador lógico: and, or, not.
- Pertenencia: Verificación mediante in o not in.
- Bandera (flag): Variable booleana usada para controlar procesos.
- Ciclo infinito: Bucle cuya condición nunca llega a False.
- Constructorbool(): Función que convierte valores a booleanos.
