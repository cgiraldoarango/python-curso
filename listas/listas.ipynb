# Listas en Python

Las listas son una de las estructuras de datos más utilizadas en Python. Permiten almacenar colecciones de elementos de forma ordenada y son fundamentales para el desarrollo de aplicaciones. Python considera las listas como estructuras de datos mutables que pueden crecer o disminuir dinámicamente.

### 1. ¿Qué es una lista?
Una lista es una estructura de datos que permite almacenar múltiples elementos en una sola variable.

```python
numeros = [10, 20, 30, 40]
```

En este caso, la variable numeros contiene cuatro elementos.

### 2. Características de las listas
Las listas en Python poseen las siguientes características:
+ Son ordenadas.
+ Son mutables.
+ Permiten elementos repetidos.
+ Pueden contener diferentes tipos de datos.
+ Son dinámicas (pueden crecer o disminuir).
+ Admiten indexación y slicing.

Ejemplo

```python
datos = [10, "Python", 3.14, True]

print(datos)
```

Las listas pueden contener elementos heterogéneos.

### 3. Restricciones de las listas
Aunque son muy flexibles, existen algunas restricciones:
#### No se puede acceder a posiciones inexistentes

```python
lista = [1, 2, 3]

print(lista[10])
```

#### El índice debe ser entero

```python
lista = [1, 2, 3]

print(lista["0"])
```

### 4. Propiedades de las listas
|Propiedad|	Descripción|
|---------|-|
|Ordenadas|	Conservan el orden de inserción|
|Mutables|	Se pueden modificar|
|Dinámicas	|Cambian de tamaño|
|Indexadas|	Permiten acceder por posición|
|Heterogéneas	|Admiten distintos tipos de datos|

### 5. Mutabilidad
Las listas son mutables, lo que significa que podemos modificar sus elementos después de crearlas.

```python
frutas = ["manzana", "pera", "uva"]

frutas[1] = "mango"

print(frutas)
```

### 6. Constructor list()
Python proporciona el constructor list() para crear listas.
#### Lista vacía

```python
lista = list()

print(lista)
```

#### Convertir una cadena

```python
letras = list("Python")

print(letras)
```

#### Convertir un rango

```python
numeros = list(range(5))

print(numeros)
```

### 7. Tipos de elementos en las listas
Las listas pueden contener:
#### Enteros

```python
[1, 2, 3]
```

#### Flotantes

```python
[3.14, 2.5]
```

#### Cadenas

```python
["hola", "mundo"]
```

#### Booleanos

```python
[True, False]
```

#### Otras listas

```python
[[1, 2], [3, 4]]
```

#### Mezcla de elementos

```python
[1, "Python", 3.14, False]
```

### 8. Tamaño de una lista
La función len() devuelve la cantidad de elementos almacenados

```python
lista = [10, 20, 30, 40]

print(len(lista))
```

### 9. Indexado de listas
#### Índices positivos

```python
numeros = [10, 20, 30, 40]

print(numeros[0])
print(numeros[2])
```

#### Índices negativos

```python
numeros = [10, 20, 30, 40]

print(numeros[-1])
print(numeros[-2])
```

### 10. Slicing de listas
Permite obtener una parte de la lista.

```python
numeros = [10, 20, 30, 40, 50]

print(numeros[1:4])
```

#### Desde el inicio

```python
print(numeros[:3])
```

#### Hasta el final

```python
print(numeros[2:])
```

### 11. Operaciones con listas
#### Pertenencia

```python
lista = [1, 2, 3]

print(2 in lista)
```

#### Repetición

```python
lista = [1, 2]

print(lista * 3)
```

#### Comparación

```python
print([1, 2] == [1, 2])
```

### 12. Concatenar listas
La forma más común es utilizar +.

```python
lista1 = [1, 2, 3]
lista2 = [4, 5, 6]

resultado = lista1 + lista2

print(resultado)
```

### 13. Comprensión de listas
Las comprensiones de listas permiten crear listas de forma compacta y eficiente.
#### Ejemplo básico

```python
cuadrados = [x**2 for x in range(10)]

print(cuadrados)
```

#### Con condición

```python
pares = [x for x in range(20) if x % 2 == 0]

print(pares)
```

#### Convertir texto a mayúsculas

```python
palabras = ["python", "java", "c#"]
resultado = [p.upper() for p in palabras]

print(resultado)
```

### 14. Métodos de las listas
#### append()
Agrega un elemento al final.

```python
lista =[1, 2]

lista.append(3)

print(lista)
```

#### extend()
Agrega múltiples elementos.

```python
lista = [1, 2]

lista.extend([3, 4])
print(lista)
```

#### insert()
Inserta un elemento en una posición determinada.

```python
lista= [1, 3]

lista.insert(1, 2)

print(lista)
```

#### remove()
Elimina la primera aparición del elemento.

```python
lista = [1, 2, 3]

lista.remove(2)

print(lista)
```

#### pop()
Elimina y retorna un elemento.

```python
lista = [10, 20, 30]

valor = lista.pop()

print(valor)
print(lista)
```

#### clear()
Vacía la lista.

```python
lista = [1, 2, 3]

lista.clear()

print(lista)
```

#### index()
Obtiene la posición de un elemento.

```python
lista = ["a", "b", "c"]

print(lista.index("b"))
```

#### count()
Cuenta ocurrencias.

```python
lista = [1, 1, 1, 2, 2]

print(lista.count(1))
```

#### reverse()
Invierte la lista.

```python
lista =[1, 2, 3]

lista.reverse()

print(lista)
```

#### sort()
Ordena la lista.

```python
lista = [5, 1, 8, 3]

lista.sort()

print(lista)
```

Orden descendente

```python
lista.sort(reverse=True)

print(lista)
```

#### copy()
Realiza una copia superficial.

```python
lista1 = [1, 2, 3]

lista2 = lista1.copy()

print(lista2)
```

### 15. Funciones útiles con listas
#### min()

```python
numeros = [10, 20, 5]

print(min(numeros))
```

#### max()

```python
print(max(numeros))
```

#### sum()

```python
print(sum(numeros))
```

#### sorted()
Retorna una nueva lista ordenada.

```python
numeros = [5, 3, 2, 7]

nueva = sorted(numeros)

print(nueva)
```

### Errores comunes
##### Modificar una posición inexistente

```python
lista = [1, 2]

lista[5] = 10
```

##### Eliminar un elemento inexistente

```python
lista.remove(100)
```

##### Confundir append() y extend()

```python
lista = [1, 2]

lista.append([3, 4])

print(lista)
```

Mientras que:

```python
lista = [1, 2]

lista.extend([3, 4])

print(lista)
```

### Resumen
Las listas son:
+ Ordenadas.
+ Mutables.
+ Dinámicas.
+ Permiten elementos repetidos.
+ Pueden contener cualquier tipo de dato.
+ Soportan indexación y slicing.
+ Incorporan numerosos métodos para inserción, eliminación, búsqueda y ordenamiento.
+ Admiten comprensiones de listas para crear colecciones de forma eficiente.


Son una de las estructuras más importantes de Python y constituyen la base para trabajar posteriormente con pilas, colas, matrices, procesamiento de datos y algoritmos más avanzados.


### Glosario
+ Lista: colección ordenada y mutable de elementos.
+ Índice: posición de un elemento dentro de la lista.
+ Slicing: extracción de una porción de la lista.
+ Mutable: que puede modificarse después de su creación.
+ Comprensión de listas: sintaxis compacta para construir listas.
+ Método: función asociada a un objeto.
+ Lista bidimensional: lista que contiene otras listas.
+ Concatenación: unión de dos o más listas.
