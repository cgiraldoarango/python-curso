# Tutorial de generación de números aleatorios en Python

El módulo random de Python proporciona funciones para trabajar con números aleatorios, seleccionar elementos y mezclar colecciones. Es ampliamente utilizado en simulaciones, videojuegos, pruebas, análisis de datos y proyectos educativos.
Antes de utilizar estas funciones debes importar el módulo:

```python
import random as rnd
```

### 1. seed()
<u>Definición</u>\
La función seed() inicializa el generador de números aleatorios con una semilla específica.\
Si utilizas la misma semilla, obtendrás siempre la misma secuencia de números aleatorios. Esto resulta muy útil para pruebas y depuración.\
<u>Sintaxis</u>

random.seed(valor)

Ejemplo

```python
import random

random.seed(10)

print(random.randint(1, 100))
print(random.randint(1, 100))
print(random.randint(1, 100))
```

Cada vez que ejecutes el programa con la semilla 10, obtendrás exactamente los mismos resultados.

<h3> Uso práctico</h3>

- Reproducir experimentos.
- Probar algoritmos.
- Simulaciones científicas.

### 2. random()

<h3> Definición </h3>

Genera un número decimal aleatorio entre 0.0 y 1.0.\
Sintaxis\
random.random()

```python
import random

numero = random.random()

print(numero)
```

Generar un número en otro rango

```python
import random

numero = random.random() * 100

print(numero)
```

#### casos de uso
- Probabilidades.
- Simulaciones.
- Juegos.

### 3. randint()

Definición\
Genera un número entero aleatorio entre dos valores, incluyendo ambos límites.\
Sintaxis\
random.randint(inicio, fin)

Ejemplo

```python
import random

numero = random.randint(1, 10)

print(numero)
```

Más ejemplos\
Simular un dado

```python
import random

dado = random.randint(1, 6)

print(dado)
```

Generar una edad

```python
edad = random.randint(18, 60)
print(edad)
```

### 4. uniform()
Definición\
Genera un número decimal aleatorio entre dos límites.\
Sintaxis\
random.uniform(a, b)

Ejemplo

```python
import random

temperatura = random.uniform(20, 30)

print(temperatura)
```

Ejemplo práctico

```python
precio = random.uniform(1000, 5000)

print(round(precio, 2))
```

Diferencia con random()\
|Función|	Rango|
|-|-|
|random()	|0.0 a 1.0|
|uniform(a,b)	|a - b|

### 5. choice()
Definición\
Selecciona un elemento aleatorio de una secuencia.\
Puede utilizarse con listas, tuplas o cadenas.\
Sintaxis\
random.choice(secuencia)

Ejemplo

```python
import random

frutas = ["Manzana", "Pera", "Banano", "Mango"]

print(random.choice(frutas))
```

Ejemplo con cadena

```python
letra = random.choice("Python")

print(letra)
```

Casos de uso
- Sorteos.
- Selección de preguntas.
- Juegos.

### 6. shuffle()
Definición\
Mezcla aleatoriamente los elementos de una lista.\
⚠️ Modifica la lista original.\
Sintaxis\
random.shuffle(lista)

Ejemplo

```python
import random

numeros = [1, 2, 3, 4, 5]

random.shuffle(numeros)

print(numeros)
```

Ejemplo práctico

```python
cartas = ["A", "K", "Q", "J"]

random.shuffle(cartas)

print(cartas)
```

### 7. sample()
Definición\
Selecciona una cantidad específica de elementos aleatorios sin repetir.\
Sintaxis\
random.sample(secuencia, cantidad)

Ejemplo

```python
import random

numeros = [1, 2, 3, 4, 5, 6, 7, 8]

muestra = random.sample(numeros, 3)

print(muestra)
```

Sorteo de estudiantes

```python
estudiantes = [
    "Ana",
    "Carlos",
    "Pedro",
    "María",
    "Luis"
]

ganadores = random.sample(estudiantes, 2)

print(ganadores)
```

Diferencia con choice()
|Función	|Resultado|
|-|-|
|choice()|	1 elemento|
|sample()	|varios elementos sin repetir|

### Resumen general
|Función|	Descripción|
|-|-|
|seed()|	Inicializa la semilla del generador|
|random()|	Decimal aleatorio entre 0.0 y 1.0|
|randint(a,b)	|Entero aleatorio entre a y b|
|uniform(a,b)|	Decimal aleatorio entre a y b|
|choice(seq)|	Selecciona un elemento aleatorio|
|shuffle(lista)|	Mezcla una lista|
|sample(seq,k)	|Obtiene k elementos sin repetir|

Ejemplo completo

```python
import random

# Semilla
random.seed(42)

# Número decimal
print("random():", random.random())

# Entero
print("randint():", random.randint(1, 10))

# Decimal entre 10 y 20
print("uniform():", random.uniform(10, 20))

# Selección de elemento
frutas = ["Manzana", "Pera", "Mango"]
print("choice():", random.choice(frutas))

# Mezclar lista
numeros = [1, 2, 3, 4, 5]
random.shuffle(numeros)
print("shuffle():", numeros)

# Muestra aleatoria
print("sample():", random.sample(numeros, 3))
```

### Posibles errores comunes
1. Usar shuffle() sobre una tupla:

```python
random.shuffle((1, 2, 3))
```

Error porque las tuplas son inmutables.

2. Solicitar más elementos de los disponibles en sample():

```python
random.sample([1, 2, 3], 5)
```

Olvidar importar el módulo:

```python
randint(1, 10)
```

Debe ser:

```python
import random

random.randint(1, 10)
```

### Glosario
- Semilla (seed): valor inicial del generador aleatorio.
- Secuencia: lista, tupla o cadena de caracteres.
- Aleatorio reproducible: secuencia que se repite usando la misma semilla.
- Muestra: subconjunto extraído de una colección.
- Barajar (shuffle): reorganizar elementos al azar.
