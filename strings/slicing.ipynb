# Slicing

El slicing es una de las características más potentes y elegantes de Python. Permite extraer, modificar o recorrer partes de secuencias como:

+ Strings (str)
+ Listas (list)
+ Tuplas (tuple)
+ Bytes (bytes)
+ Rangos (range)
+ Arrays de NumPy (con algunas extensiones)

La documentación oficial y los materiales de referencia que tienes en este cuaderno lo presentan como una operación fundamental para trabajar con secuencias.

### 1. Sintaxis básica

secuencia[inicio:fin]

Python devuelve los elementos desde inicio hasta fin - 1.

```python
texto = "Python"

print(texto[0:2])
```

Se toma:

P y t h o n 


0 1 2 3 4 5

Desde la posición 0 hasta antes de la 2.
