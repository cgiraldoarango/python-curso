El slicing es una de las características más potentes y elegantes de Python. Permite extraer, modificar o recorrer partes de secuencias como:

+ Strings (str)
+ Listas (list)
+ Tuplas (tuple)
+ Bytes (bytes)
+ Rangos (range)
+ Arrays de NumPy (con algunas extensiones)

La documentación oficial y los materiales de referencia que tienes en este cuaderno lo presentan como una operación fundamental para trabajar con secuencias.

### 1. Sintaxis básica

secuencia[inicio:fin]

Python devuelve los elementos desde inicio hasta fin - 1.

```python
texto = "Python"

print(texto[0:2])
```

Se toma:

P y t h o n 


0 1 2 3 4 5

Desde la posición 0 hasta antes de la 2.

### 2. Regla más importante

✅ El inicio se incluye
✅ El final se excluye

```python
texto = "Python"

print(texto[2:5])
```

## 3. Omitir el inicio

Si no colocas el inicio:

```python
texto[:4]
```

### 4. Omitir el final

```python
texto[3:]
```

### 5. Obtener toda la secuencia

```python
texto[:]
```

En listas suele usarse para crear una copia superficial:

```python
lista = [1,2,3]
lista2 = lista[:]
```

### 6. Índices negativos

Python permite contar desde el final.

```python
texto = "Python"
```

* P  y  t  h  o  n 
* 0 1 2 3 4 5 
* -6 -5 -4 -3 -2 -1

Último carácter

```python
texto[-1]
```

Penúltimo

```python
texto[-2]
```

### 7. Slicing con índices negativos

```python
texto[-4:-1]
```

### 8. Tercer parámetro: paso

secuencia[inicio:fin:paso]

Saltar de dos en dos

```python
texto = "Python"

print(texto[0:6:2])
```

Impares

```python
texto[1:6:2]
```

### 9. Invertir una cadena

```python
texto[::-1]
```

### 10. Paso negativo

```python
texto[5:1:-1]
```

Significa
* empieza en 5
* llega hasta antes de 1
* retrocede de uno en uno

### 11. Slicing sobre listas

```python
numeros = [10, 20, 30, 40, 50]
```

Obtener una parte

```python
numeros[1:4]
```

Primeros tres

```python
numeros[:3]
```

Últimos tres

```python
numeros[-3:]
```

12. Slicing sobre tuplas

```python
datos = (1, 2, 3, 4, 5)
datos[1:4]
```

### 13. Modificar listas usando slicing

Una diferencia importante:
Los strings son inmutables.
Las listas son mutables.

```python
lista = [1, 2, 3, 4, 5]

lista[1:4] = [20, 30, 40]

print(lista)
```

Eliminar elementos

```python
lista = [1, 2, 3, 4, 5]

lista[1:4] = []

print(lista)
```

### 14. Error común: Off-by-one

```python
texto[0:5]
```

### 15. Ventaja matemática del slicing

Una propiedad muy importante:

```python
s= "python"
s[:3] + s[3:] == s
```

### 16. Copias de listas

```python
original = [1, 2, 3]

copia = original[:]
```

### 17. Slicing multidimensional

Si tienes una lista de listas:

```python
matriz = [
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
```

Primera fila:

```python
matriz[0][:]
```

### 18. Patrones profesionales frecuentes

Clonar listas

```python
copia = lista[:]
```

Obtener los N primeros

```python
top5 = datos[:5]
```

Obtener los N últimos

```python
ultimos5 = datos[-5:]
```

Invertir

```python
datos[::-1]
```

Elementos pares

```python
datos[::2]
```

Elementos impares

```python
datos[1::2]
```

Saltar cada 3 elementos

```python
datos[::3]
```

### Resumen visual

* cadena = "PYTHON"

### Errores comunes

1. Olvidar que el final no se incluye

```python
texto[0:5]
```

2. Modificar strings

```python
texto[0] = "X"
```

3. Confundir copia con referencia

```python
b= "Cali"
a = "Bucaramanga"
b = a
```

No copia.

```python
b = a[:]
b
```

Sí copia superficialmente.

4. Usar índices fuera de rango

```python
texto[50]
```

### Glosario

+ Slice: subconjunto de una secuencia.
+ Start: índice inicial.
+ Stop: índice final (excluido).
+ Step: salto entre elementos.
+ Shallow copy: copia superficial.
+ Immutable: objeto que no puede modificarse.
+ Sequence: estructura indexable (string, lista, tupla, etc.).
