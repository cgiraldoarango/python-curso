# Métodos

Los strings (str) son uno de los tipos de datos más utilizados en Python. Permiten almacenar texto y disponen de numerosos métodos para transformarlo, validarlo y analizarlo.

### 1. upper()
Convierte todos los caracteres alfabéticos a mayúsculas.

```python
texto = "hola mundo"
resultado = texto.upper()

print(resultado)
```

### 2. lower()
Convierte todos los caracteres alfabéticos a minúsculas.

```python
texto = "HOLA MUNDO"
resultado = texto.lower()

print(resultado)
```

### 3. capitalize()
Convierte la primera letra a mayúscula y el resto a minúsculas.

```python
texto = "hOLA mUNDO"
print(texto.capitalize())
```

### 4. title()
Convierte en mayúscula la primera letra de cada palabra.

```python
texto = "python para principiantes"
print(texto.title())
```

### 5. strip()
Elimina espacios al inicio y al final de la cadena

```python
texto = "   Hola Python   "
print(texto.strip())
```

### 6. replace()
Reemplaza una subcadena por otra

```python
texto = "Me gusta Java"
print(texto.replace("Java", "Python"))
```

También se puede limitar:

```python
texto = "hola hola hola"

print(texto.replace("hola", "adios", 2))
```

### 7. split()
Divide un string en una lista

```python
texto = "Juan,Pedro,Ana"
nombres = texto.split(",")

print(nombres)
```

Por espacios:

```python
texto = "Python es genial"
print(texto.split())
```

### 8. join()
Une los elementos de una lista usando un separador.

```python
nombres = ["Juan", "Pedro", "Ana"]

resultado = "-".join(nombres)

print(resultado)
```

### 9. find()
Busca la primera aparición de una subcadena. Devuelve -1 si no existe.

```python
texto = "Python es poderoso"

print(texto.find("es"))
```

Si no existe:

```python
print(texto.find("Java"))
```

### 10. count()
Cuenta cuántas veces aparece una cadena.

```python
texto = "hola hola hola"

print(texto.count("hola"))
```

### 11. startswith()
Verifica si el string comienza con cierto texto.

```python
archivo = "reporte.pdf"

print(archivo.startswith("rep"))
```

### 12. endswith()
Verifica si termina con cierto texto.

```python
archivo = "reporte.pdf"

print(archivo.endswith(".pdf"))
```

### 13. isalpha()
Devuelve True si contiene solamente letras.

```python
texto = "Python"

print(texto.isalpha())
```

```python
texto = "Python3"

print(texto.isalpha())
```

### 14. isdigit()
Comprueba si contiene únicamente números.

```python
texto = "123456"

print(texto.isdigit())
```

```python
texto = "123A"

print(texto.isdigit())
```

### 15. isalnum()
Comprueba si contiene solo letras y números.

```python
texto = "Python123"

print(texto.isalnum())
```

### 16. isspace()
Comprueba si contiene únicamente espacios.

```python
texto = "     "

print(texto.isspace())
```

### 17. center()
Centra un texto dentro de un ancho determinado.

```python
texto = "Python"

print(texto.center(20, "-"))
```

### 18. ljust()
Alinea a la izquierda.

```python
texto = "Python"

print(texto.ljust(15, "."))
```

### 19. rjust()
Alinea a la derecha.

```python
texto = "Python"

print(texto.rjust(15, "."))
```

### 20. zfill()
Rellena con ceros a la izquierda

```python
numero = "42"

print(numero.zfill(6))
```

### 21. format()
Permite insertar valores dentro de una cadena

```python
nombre = "Carlos"
edad = 30

texto = "Me llamo {} y tengo {} años".format(nombre, edad)

print(texto)
```

### 22. f-strings (Recomendado)
La forma moderna de formatear cadenas.

```python
nombre = "Carlos"
edad = 30

print(f"Me llamo {nombre} y tengo {edad} años")
```

### 23. len() (función relacionada con strings)
Aunque no es un método, es fundamental para trabajar con cadenas.

```python
texto = "Python"

print(len(texto))
```

### 24. Indexación y slicing
Obtener un carácter

```python
texto = "Python"

print(texto[0])
```

Obtener una parte

```python
texto = "Python"

print(texto[1:4])
```

Último carácter

```python
print(texto[-1])
```

### Ejemplo práctico completo

```python
nombre = "   cArLos giRalDo   "

nombre = nombre.strip()
nombre = nombre.title()

print(nombre)

print("Longitud:", len(nombre))
print("Empieza por C:", nombre.startswith("C"))
print("Contiene 'Giraldo':", "Giraldo" in nombre)
print("Número de letras o:", nombre.lower().count("o"))
```

### Errores comunes

Pensar que los métodos modifican el string original.

```python
texto = "hola"
texto.upper()

print(texto)
```

Usar find() esperando un error cuando no existe.

```python
texto.find("Java")
```

Intentar modificar un carácter directamente.

```python
texto = "Python"
texto[0] = "J"
```

Los strings son inmutables

### Glosario
1. String (str): cadena de texto.
2. Método: función asociada a un objeto.
3. Subcadena: parte de un string.
4. Índice: posición de un carácter.
5. Slicing: extracción de una porción de texto.
6. Inmutable: objeto que no puede modificarse directamente.
