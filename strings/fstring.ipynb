# f.string

Los f-strings (formatted string literals) son una forma moderna y eficiente de construir cadenas de texto en Python. Fueron introducidos en Python 3.6 y permiten insertar variables y expresiones directamente dentro de una cadena usando la letra f antes de las comillas.

```python
nombre= "carlos"
apellido = "Giraldo"
print(f"El nombre es {nombre} y el apellido es {apellido}")
```

### Cómo funciona
Todo lo que está entre llaves será evaluado

```python
a = 10
b = 20

print(f"La suma es {a + b}")
```

### Formateo de números

#### Decimales

```python
pi = 3.14159265

print(f"{pi:.2f}")
```

#### Explicación

* .2f = mostrar 2 decimales
* .4f = mostrar 4 decimales

```python
pi = 3.14159265
print(f"{pi:,.4f}")
```

### Porcentajes

```python
porcentaje = 0.875

print(f"{porcentaje:.2%}")
```

### Separador de miles

```python
numero = 123456789
print(f"{numero:,}")
```

### Alineación de texto

#### Izquierda

```python
nombre = "Pedro"
print(f"|{nombre:<10}|")
```

#### Derecha

```python
print(f"|{nombre:>10}|")
```

#### Centrado

```python
print(f"|{nombre:^10}|")
```

#### Relleno con caracteres

```python
numero = 25
print(f"{numero:0>5}")
```

#### Evaluar expresiones

```python
nombre = "Carlos"
print(f"Mayúsculas: {nombre.upper()}")
```

#### Llamar métodos

```python
texto = "python"

print(f"{texto.capitalize()}")
```

#### Mostar nombre y valor de una variable

```python
edad = 25

print(f"{edad=}")
```

##### Otro ejemplo

```python
a = 5
b = 8

print(f"{a=}, {b=}, {a+b=}")
```

#### Uso con fechas

```python
from datetime import datetime

fecha = datetime.now()

print(f"{fecha:%d/%m/%Y}")
```

### Uso en FASTAPI

```python
from fastapi import FastAPI

app = FastAPI()

@app.get("/usuarios/{id}")
def obtener_usuario(id: int):
    return {
        "mensaje": f"Consultando usuario con ID {id}"
    }
```

#### Ejemplo práctico

```python
nombre = "María"
edad = 28
salario = 3567.899

mensaje = (
    f"Empleado: {nombre}\n"
    f"Edad: {edad}\n"
    f"Salario: ${salario:,.2f}"
)

print(mensaje)
```

### Glosario

Término	Descripción
1. f-string	Cadena formateada que comienza con f
2. Placeholder	Expresión dentro de {}
3. Formateo	Modificación visual de un valor
4. Expresión	Código evaluado dentro de las llaves
5. Debugging	Proceso de depuración de errores
6. Specifier	Instrucción de formato (.2f, >10, %)

#### Resumen

```python
nombre = "Ana"
precio = 1234.567

print(f"Hola {nombre}")
print(f"{precio:.2f}")
print(f"{precio:,.2f}")
print(f"{nombre.upper()}")
print(f"{precio=}")
```
